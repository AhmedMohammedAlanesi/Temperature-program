```python
def celsius_to_fahrenheit(celsius):
    """يحول درجة الحرارة من مئوي إلى فهرنهايت."""
    fahrenheit = (celsius * 9/5) + 32
    return fahrenheit

# مثال على الاستخدام:
celsius_input = input("ادخل درجة الحرارة بالمئوية: ")
try:
    celsius_temp = float(celsius_input)
    fahrenheit_temp = celsius_to_fahrenheit(celsius_temp)
    print(f"درجة الحرارة {celsius_temp} مئوية تعادل {fahrenheit_temp} فهرنهايت.")
except ValueError:
    print("ادخال غير صالح. يرجى ادخال رقم.")
```
